# python과 데이터 베이스 연동
- 내장 DB 연동 : SQLite

```python
import sqlite3

# isolation_level : 트랜잭션
# autoCommit 중지
conn = sqlite3.connect("./test.db",isolation_level=None)

# sql 을 데이터베이스에 전달하고 결과를 받아오는 객체
cursor = conn.cursor()
```

```python
# 테이블 생성
sql = """create table if not exists users(
id integer primary key, 
username text,
email text,
phone text,
website text,
regdate text
)"""

cursor.execute(sql)
```

```python
from datetime import datetime
now = datetime.now()
nowDatetime = now.strftime("%Y-%m-%d %H:%M:%S")
nowDatetime
```

```python
# insert 구문 : 튜플 형태로
sql = "insert into users(id,username,email,phone,website,regdate)" \
"values(1,'kim','asd@asd.asd','010-1234-5678','kim.com',?)"

cursor.execute(sql,(nowDatetime,))
```

```python
sql = "insert into users(id,username,email,phone,website,regdate)" \
"values(?,?,?,?,?,?)"

cursor.execute(sql,(2,'hong','asd2@asd.asd','010-1234-5678','hong.com',nowDatetime,))
```

```python
user_list = [
    (3,'park','asd3@asd.asd','010-1234-5678','park.com',nowDatetime,),
    (4,'lee','asd4@asd.asd','010-1234-5678','lee.com',nowDatetime,),
    (5,'jo','asd5@asd.asd','010-1234-5678','jo.com',nowDatetime,),
]

cursor.executemany(sql,user_list)
```

```python
# 데이터 조회

sql = "select * from users"

cursor.execute(sql)
```

```python
cursor.fetchone()
```

```python
cursor.fetchmany(size=3)
```

```python
cursor.fetchall()
```

```python
sql = "select * from users order by id desc"

cursor.execute(sql)

for row in cursor.fetchall():
    print(row)
```

```python
sql = "select * from users where id = ?"

cursor.execute(sql,(4,))

for row in cursor.fetchall():
    print(row)
```

```python
sql = "select * from users where id in (?,?)"

cursor.execute(sql,(2,5))
cursor.fetchall()
```

```python
sql = "update users set username = ? where id = ?"
cursor.execute(sql,('young',5,))
```

```python
sql = "select * from users where id = 5"

cursor.execute(sql)
cursor.fetchone()
```

```python
# 삭제

sql = "delete from users where id = ?"
cursor.execute(sql, (5,))
```
