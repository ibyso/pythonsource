```python
from faker import Faker
fake = Faker(locale="ko_KR")
Faker.seed(4321)

print(fake.name())
print(fake.email())
print(fake.address())
print(fake.ssn())
print(fake.phone_number())
```

```python
# maru_member 50명 임의의 데이터를 이용해 삽입
# name, birth, phone, joined_at
from datetime import datetime
sql = "INSERT INTO MARU_MEMBER (MEMBER_ID, NAME, BIRTH, PHONE, JOINED_AT) VALUES(:member_id , :name, :birth, :phone, :joined_at)"
fake = Faker(locale="ko_KR")
data = []

for i in range(50) :
    data.append([i+1,fake.name(),fake.date_of_birth(minimum_age=20,maximum_age=70).strftime("%Y-%m-%d"),fake.phone_number(),fake.date_time_between(start_date="-3y",end_date="now")])

with oracledb.connect(user='python_user',password=password,dsn="localhost/xe") as conn :
    with conn.cursor() as cursor :
        cursor.executemany(sql,data)
        conn.commit()
```
