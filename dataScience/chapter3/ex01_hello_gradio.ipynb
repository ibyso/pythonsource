## Hello Gradio - 첫 인터페이스 만들기
- `gr.Interface`로 가장 기본적인 웹 인터페이스를 만드는 흐름
- "함수 하나 -> 입력창 하나 -> 출력창 하나" 라는 gradio의 기본 구조 이해
- uv pip install gradio
- 설치 후 버전확인 : python -c "import gradio as gr; print(gr.__version__)"

```python
import gradio as gr # gradio 라이브러리

def greet(name):
    return f"안녕하세요, {name}님! 반갑습니다."

demo = gr.Interface(
    fn=greet,
    inputs="text",  # inputs="number" 입력창 모양이 변경 str -> float
    outputs="text",
    title="첫 gradio 인터페이스",
    description="이름을 입력하면 인사말을 보여줍니다.",
)
```

```python
demo.launch()
demo.close()
```
