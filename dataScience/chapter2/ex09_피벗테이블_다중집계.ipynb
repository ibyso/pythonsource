## 피벗테이블과 다중 집계 함수

- `pivot_table()`로 두 개의 범주형 변수(지역x카테고리)를 축으로 하는 교차표
- 하나의 피벗테이블 안에서 평균/합계/개수 등 여러 집계 함수를 동시에 적용

```python
import os
from dotenv import load_dotenv
import pandas as pd
import oracledb

load_dotenv()

USER = os.getenv("ORACLE_USER")
PASSWORD = os.getenv("ORACLE_PASSWORD")
DSN = os.getenv("ORACLE_DSN")

with oracledb.connect(user=USER,password=PASSWORD,dsn=DSN) as conn :
    df = pd.read_sql("select * from delivery_orders", conn)

df["ORDER_DATE"] = pd.to_datetime(df["ORDER_DATE"])
print(f"전체 주문 데이터 : {len(df)}건")
df.head()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import matplotlib as mpl

mpl.rc('font',family='Malgun Gothic')
mpl.rc('axes',unicode_minus=False)
```

```python
import numpy as np

pivot_price =pd.pivot_table(
    df,
    values="PRICE",
    index="REGION",
    columns="CATEGORY",
    aggfunc="mean",
    fill_value=0,
    margins=True, # 총계 행/열 자동으로 추가
    margins_name="전체평균", # 원래 ALL인데 전체평균
)

print(pivot_price.round(0))
```

```python
pivot_multi =pd.pivot_table(
    df,
    values="PRICE",
    index="REGION",
    columns="CATEGORY",
    aggfunc=["mean","count"], # 평균과 건수를 동시에
    fill_value=0
)

print(pivot_multi)
```

```python
import seaborn as sns

plt.figure(figsize=(8,5))
# annot = True : 칸 안에 숫자 표기
# 전체평균 삭제하여 큰 값에 대한 무조건적인 대표값으로 취급하는 왜곡현상을 막는다.
sns.heatmap(pivot_price.drop("전체평균",axis=0).drop("전체평균",axis=1),annot=True,fmt=".0f",cmap="YlOrRd")
plt.title("지역 x 카테고리 평균 가격 히트맵")
plt.show()
```
